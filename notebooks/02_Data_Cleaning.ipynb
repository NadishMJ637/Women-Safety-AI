```python
import ast

def decode_tweet(tweet):
    try:
        if isinstance(tweet, str) and tweet.startswith("b'"):
            return ast.literal_eval(tweet).decode("utf-8", errors="ignore")
        return tweet
    except:
        return tweet
```

```python
df["Tweets"] = df["Tweets"].apply(decode_tweet)
```
